## Problema 1 — Calculadora General de Permutaciones y $k$-Permutaciones

### 1. Descripción Matemática

Sea $S$ un conjunto de $n$ objetos **distintos**. Una **$r$-permutación** de $S$ es una secuencia ordenada de $r$ elementos distintos de $S$, donde el **orden importa**.  

Formalmente, si $0 \leq r \leq n$, el número de $r$-permutaciones de un conjunto de $n$ elementos es:

$$P(n, r) = \frac{n!}{(n-r)!} = n \cdot (n-1) \cdot (n-2) \cdots (n-r+1)$$

**Casos especiales:**
- Cuando $r = n$: $P(n, n) = n!$ (todas las permutaciones del conjunto completo).
- Cuando $r = 0$: $P(n, 0) = 1$ (solo existe la secuencia vacía, convenio $0! = 1$).

### 2. Fórmula Combinatoria

$$\boxed{P(n, r) = \frac{n!}{(n-r)!}}$$

### 3. Justificación del Algoritmo

El algoritmo sigue el **principio de la multiplicación**:  
- Para la primera posición hay $n$ elecciones.  
- Para la segunda (ya ocupada una) hay $n-1$ elecciones.  
- $\ldots$  
- Para la $r$-ésima posición hay $n-r+1$ elecciones.  

El producto de estas elecciones da exactamente $P(n, r)$.  
Computacionalmente, `math.factorial` garantiza precisión entera exacta para cualquier $n$ razonable.

### 4. Código Funcional

```python
import math

def factorial(n: int) -> int:
    """Calcula n! usando la función integrada de Python para máxima precisión."""
    return math.factorial(n)

def permutacion(n: int, r: int) -> int:
    """
    Calcula P(n, r) = n! / (n-r)!
    
    Parámetros:
        n (int): tamaño del conjunto total. Debe ser un entero no negativo.
        r (int): número de elementos a ordenar. Debe cumplir 0 <= r <= n.
    
    Retorna:
        int: número de r-permutaciones del conjunto.
    
    Lanza:
        ValueError: si n o r no cumplen las condiciones matemáticas.
        TypeError: si n o r no son enteros.
    """
    # --- Validación de tipo ---
    if not isinstance(n, int) or not isinstance(r, int):
        raise TypeError(f"n y r deben ser enteros. Recibido: n={type(n).__name__}, r={type(r).__name__}")
    
    # --- Validación matemática ---
    if n < 0:
        raise ValueError(f"n debe ser un entero no negativo. Recibido: n={n}")
    if r < 0:
        raise ValueError(f"r debe ser un entero no negativo. Recibido: r={r}")
    if r > n:
        raise ValueError(f"r no puede ser mayor que n. Condición violada: {r} > {n}")
    
    resultado = factorial(n) // factorial(n - r)
    return resultado


def mostrar_permutacion(n: int, r: int) -> None:
    """Muestra el procedimiento matemático paso a paso de P(n, r)."""
    print(f"\n{'='*52}")
    print(f"  Calculando P({n}, {r})")
    print(f"{'='*52}")
    print(f"  Fórmula:   P(n, r) = n! / (n - r)!")
    print(f"  Sustitución: P({n}, {r}) = {n}! / ({n} - {r})! = {n}! / {n-r}!")
    print(f"  n!  = {n}! = {factorial(n):,}")
    print(f"  (n-r)! = {n-r}! = {factorial(n-r):,}")
    
    # Mostrar el producto descendente
    if r > 0:
        terminos = list(range(n, n - r, -1))
        producto_str = " × ".join(str(t) for t in terminos)
        print(f"  Producto:  {producto_str} = {permutacion(n, r):,}")
    else:
        print(f"  Producto:  (vacío — solo existe la secuencia vacía)")
    
    resultado = permutacion(n, r)
    print(f"  ─────────────────────────────────────")
    print(f"  ✓ P({n}, {r}) = {resultado:,}")
    print(f"{'='*52}\n")

print("✅ Módulo de permutaciones cargado correctamente.")
```

### 5. Pruebas — Casos Requeridos y Adicionales

```python
# ── Prueba 1: P(10, 3) ──────────────────────────────────────
mostrar_permutacion(10, 3)

# ── Prueba 2: P(20, 5) ──────────────────────────────────────
mostrar_permutacion(20, 5)
```

```python
# ── Prueba 3: Caso especial r = 0  (único resultado: secuencia vacía) ──
mostrar_permutacion(7, 0)

# ── Prueba 4: Caso especial r = n  (permutación total) ──────
mostrar_permutacion(5, 5)
```

```python
# ── Prueba 5: Valor grande para verificar escalabilidad ─────
mostrar_permutacion(52, 5)  # manos de póker: 52 cartas, elige 5 en orden
```

```python
# ── Prueba 6 (validación de error): r > n — debe lanzar ValueError ──
print("Prueba de error: P(5, 8) — r > n")
try:
    permutacion(5, 8)
except ValueError as e:
    print(f"  ✓ ValueError capturado correctamente: {e}")

print()

# ── Prueba 7 (validación de error): n negativo ───────────────
print("Prueba de error: P(-3, 2) — n negativo")
try:
    permutacion(-3, 2)
except ValueError as e:
    print(f"  ✓ ValueError capturado correctamente: {e}")

print()

# ── Prueba 8 (validación de error): tipo inválido ────────────
print("Prueba de error: P(4.5, 2) — n no es entero")
try:
    permutacion(4.5, 2)
except TypeError as e:
    print(f"  ✓ TypeError capturado correctamente: {e}")
```

### 6. Eficiencia del Algoritmo

La función `math.factorial` de Python usa la biblioteca GMP en el backend, con complejidad aproximada $O(n \log n \log \log n)$ para $n$ grande (algoritmo de divide y vencerás). Para los valores de $n$ típicos en combinatoria (hasta algunos miles), la ejecución es prácticamente instantánea. El cómputo de $P(n, r)$ como cociente de dos factoriales introduce una multiplicación adicional de enteros grandes, manteniendo la precisión entera exacta — sin errores de punto flotante.

## Problema 9 — Coeficientes Multinomiales y Palabras con Letras Repetidas

### 1. Descripción Matemática

Dado un multiconjunto de $n$ objetos donde hay $k$ grupos de objetos indistinguibles entre sí, con $n_1$ objetos del tipo 1, $n_2$ del tipo 2, $\ldots$, $n_k$ del tipo $k$, de modo que:

$$n = n_1 + n_2 + \cdots + n_k$$

el número de permutaciones **distinguibles** de esos $n$ objetos es el **coeficiente multinomial**:

$$\binom{n}{n_1, n_2, \ldots, n_k} = \frac{n!}{n_1! \cdot n_2! \cdots n_k!}$$

**Intuición:** se divide entre los factoriales de las repeticiones para eliminar las permutaciones que, aunque distintas en posición, producen el mismo arreglo visible.

**Caso particular — coeficiente binomial:** cuando $k = 2$, recuperamos la expresión familiar:
$$\binom{n}{r} = \frac{n!}{r!\,(n-r)!}$$

### 2. Fórmula Combinatoria

$$\boxed{M(n_1, n_2, \ldots, n_k) = \frac{(n_1 + n_2 + \cdots + n_k)!}{n_1! \cdot n_2! \cdots n_k!}}$$

### 3. Justificación del Algoritmo

1. Si la entrada es una **palabra**, se cuenta la frecuencia de cada carácter mediante `collections.Counter`.  
2. Si la entrada es una **lista de cantidades**, se usa directamente.  
3. Se calcula $n = \sum n_i$ y se aplica la fórmula.  
4. La división es siempre exacta (resultado entero) — se usa división entera `//` para evitar artefactos de punto flotante.

### 4. Código Funcional

```python
import math
from collections import Counter

def multinomial(frecuencias: list[int]) -> int:
    """
    Calcula el coeficiente multinomial n! / (n1! * n2! * ... * nk!)
    
    Parámetros:
        frecuencias (list[int]): lista [n1, n2, ..., nk] con las cantidades
                                  de cada tipo de objeto. Todos deben ser >= 1.
    
    Retorna:
        int: número de permutaciones distinguibles.
    
    Lanza:
        ValueError: si algún elemento no es un entero positivo.
        TypeError: si frecuencias no es una lista de enteros.
    """
    if not frecuencias:
        raise ValueError("La lista de frecuencias no puede estar vacía.")
    
    for i, f in enumerate(frecuencias):
        if not isinstance(f, int):
            raise TypeError(f"Todos los elementos deben ser enteros. Elemento {i}: {type(f).__name__}")
        if f < 1:
            raise ValueError(f"Cada frecuencia debe ser un entero positivo (>= 1). Recibido: n{i+1}={f}")
    
    n = sum(frecuencias)
    denominador = 1
    for f in frecuencias:
        denominador *= math.factorial(f)
    
    return math.factorial(n) // denominador


def palabras_desde_texto(texto: str) -> int:
    """
    Cuenta las palabras distinguibles formadas con las letras de `texto`.
    Ignora espacios y convierte a mayúsculas para normalización.
    """
    texto_limpio = texto.upper().replace(" ", "")
    if not texto_limpio:
        raise ValueError("El texto no puede estar vacío o contener solo espacios.")
    conteo = Counter(texto_limpio)
    return multinomial(list(conteo.values()))


def mostrar_multinomial_texto(texto: str) -> None:
    """Muestra el procedimiento completo para una palabra dada."""
    texto_limpio = texto.upper().replace(" ", "")
    conteo = Counter(texto_limpio)
    n = len(texto_limpio)
    
    print(f"\n{'='*56}")
    print(f"  Palabras distinguibles con: \"{texto.upper()}\"")
    print(f"{'='*56}")
    print(f"  Letras totales (n): {n}")
    print(f"  Frecuencias por letra:")
    for letra, freq in sorted(conteo.items()):
        print(f"    {letra}: {freq} (contribuye {freq}! = {math.factorial(freq):,} al denominador)")
    
    denom_str = " × ".join(f"{freq}!" for _, freq in sorted(conteo.items()))
    denom_val = 1
    for freq in conteo.values():
        denom_val *= math.factorial(freq)
    
    print(f"  Fórmula aplicada:")
    print(f"    M = {n}! / ({denom_str})")
    print(f"      = {math.factorial(n):,} / {denom_val:,}")
    resultado = multinomial(list(conteo.values()))
    print(f"  ──────────────────────────────────────────")
    print(f"  ✓ Palabras distintas: {resultado:,}")
    print(f"{'='*56}\n")


def mostrar_multinomial_lista(etiquetas: list[str], cantidades: list[int]) -> None:
    """Muestra el procedimiento para una lista de cantidades con etiquetas."""
    n = sum(cantidades)
    print(f"\n{'='*56}")
    print(f"  Cadenas distinguibles con cantidades dadas")
    print(f"{'='*56}")
    print(f"  Objetos:")
    for etq, cnt in zip(etiquetas, cantidades):
        print(f"    {etq}: {cnt} (contribuye {cnt}! = {math.factorial(cnt):,} al denominador)")
    
    denom_str = " × ".join(f"{c}!" for c in cantidades)
    denom_val = 1
    for c in cantidades:
        denom_val *= math.factorial(c)
    
    print(f"  Total n = {' + '.join(str(c) for c in cantidades)} = {n}")
    print(f"  Fórmula aplicada:")
    print(f"    M = {n}! / ({denom_str})")
    print(f"      = {math.factorial(n):,} / {denom_val:,}")
    resultado = multinomial(cantidades)
    print(f"  ──────────────────────────────────────────")
    print(f"  ✓ Cadenas distintas: {resultado:,}")
    print(f"{'='*56}\n")

print("✅ Módulo de coeficientes multinomiales cargado correctamente.")
```

### 5. Pruebas — Casos Requeridos y Adicionales

```python
# ── Prueba 1 (requerida): BANANA ─────────────────────────────
mostrar_multinomial_texto("BANANA")
```

```python
# ── Prueba 2 (requerida): 4A, 3B, 2C ─────────────────────────
mostrar_multinomial_lista(["A", "B", "C"], [4, 3, 2])
```

```python
# ── Prueba 3: MISSISSIPPI (clásico de combinatoria) ───────────
mostrar_multinomial_texto("MISSISSIPPI")
```

```python
# ── Prueba 4: Caso especial — todos los caracteres distintos ──
# Para una cadena sin repeticiones, el resultado es n!
mostrar_multinomial_texto("ABCDE")
print(f"  Verificación: 5! = {math.factorial(5)} ✓")
```

```python
# ── Prueba 5: Caso especial — todos los caracteres iguales ────
# Para n copias del mismo carácter, solo existe 1 arreglo
mostrar_multinomial_texto("AAAAAAA")
print(f"  Verificación: 7!/7! = 1 ✓")
```

```python
# ── Prueba 6 (error): frecuencia = 0 — debe lanzar ValueError ─
print("Prueba de error: frecuencia 0 en la lista")
try:
    multinomial([3, 0, 2])
except ValueError as e:
    print(f"  ✓ ValueError capturado: {e}")

print()

# ── Prueba 7 (error): texto vacío ─────────────────────────────
print("Prueba de error: texto vacío")
try:
    palabras_desde_texto("   ")
except ValueError as e:
    print(f"  ✓ ValueError capturado: {e}")

print()

# ── Prueba 8 (error): tipo no entero en lista ─────────────────
print("Prueba de error: tipo flotante en lista")
try:
    multinomial([3, 2.5, 1])
except TypeError as e:
    print(f"  ✓ TypeError capturado: {e}")
```

### 6. Eficiencia del Algoritmo

El algoritmo es $O(k + n \log n \log \log n)$, donde:
- $k$ es el número de grupos distintos (dominado por el conteo con `Counter`, que es $O(n)$),
- el término de $n!$ domina asintóticamente.

El uso de división entera exacta (`//`) evita cualquier pérdida de precisión. Para palabras de decenas de caracteres, la ejecución es imperceptible. La complejidad espacial es $O(k)$ para almacenar las frecuencias.
